```python
import pandas as pd

DATASET_URL = (
    "https://storage.googleapis.com/"
    "download.tensorflow.org/data/creditcard.csv"
)

df = pd.read_csv(DATASET_URL)

print(f"Dimensões do dataset: {df.shape[0]:,} linhas e {df.shape[1]} colunas")

display(df.head())
```

```python
print("Colunas:")
print(df.columns.tolist())

print("\nTipos de dados:")
display(df.dtypes.to_frame(name="tipo"))

print("\nValores ausentes por coluna:")
display(df.isna().sum().to_frame(name="valores_ausentes"))

print(f"\nLinhas duplicadas: {df.duplicated().sum():,}")

print("\nDistribuição da variável Class:")
display(
    df["Class"]
    .value_counts()
    .rename(index={0: "Normal", 1: "Fraude"})
    .to_frame(name="quantidade")
)
```

```python
proporcao = df["Class"].value_counts(normalize=True).sort_index() * 100

print(f"Transações normais: {proporcao.get(0, 0):.4f}%")
print(f"Transações fraudulentas: {proporcao.get(1, 0):.4f}%")
```

```python
# Linhas envolvidas em grupos de duplicatas exatas
duplicadas_todas = df.duplicated(keep=False)

print(
    "Linhas pertencentes a grupos de duplicatas:",
    f"{duplicadas_todas.sum():,}"
)

print(
    "Ocorrências repetidas após a primeira:",
    f"{df.duplicated().sum():,}"
)

print("\nClasses nas linhas envolvidas em duplicatas:")
display(
    df.loc[duplicadas_todas, "Class"]
    .value_counts()
    .rename(index={0: "Normal", 1: "Fraude"})
    .to_frame(name="quantidade")
)

print("\nExemplos para inspeção:")
display(df.loc[duplicadas_todas].head(10))
```

```python
feature_cols = [col for col in df.columns if col != "Class"]

rotulos_por_atributos = (
    df.groupby(feature_cols, dropna=False)["Class"]
    .nunique()
)

conflitos = rotulos_por_atributos[rotulos_por_atributos > 1]

print(
    "Grupos de atributos idênticos com rótulos diferentes:",
    f"{len(conflitos):,}"
)
```

```python
# Identifica ocorrências repetidas após a primeira
mascara_duplicatas = df.duplicated(keep="first")

# Analisa somente as linhas que seriam removidas
duplicatas_removidas = df.loc[mascara_duplicatas].copy()

print(f"Linhas que seriam removidas: {len(duplicatas_removidas):,}")

print("\nDistribuição das classes nas linhas removidas:")
display(
    duplicatas_removidas["Class"]
    .value_counts()
    .rename(index={0: "Normal", 1: "Fraude"})
    .to_frame(name="quantidade")
)

print("\nDistribuição percentual das classes nas linhas removidas:")
display(
    (
        duplicatas_removidas["Class"]
        .value_counts(normalize=True)
        .mul(100)
        .rename(index={0: "Normal", 1: "Fraude"})
        .to_frame(name="percentual")
    )
)
```

```python
# Cenário A: dataset original
df_original = df.copy()

# Cenário B: somente para análise de sensibilidade
df_sem_duplicatas = df.drop_duplicates().copy()

for nome, dados in [
    ("Original", df_original),
    ("Sem duplicatas exatas", df_sem_duplicatas),
]:
    fraudes = int((dados["Class"] == 1).sum())
    total = len(dados)

    print(
        f"{nome}: {total:,} linhas | "
        f"{fraudes:,} fraudes | "
        f"{fraudes / total * 100:.4f}% de fraude"
    )
```

```python
# Verificações básicas de qualidade
expected_columns = (
    ["Time", "Amount"]
    + [f"V{i}" for i in range(1, 29)]
    + ["Class"]
)

missing_columns = set(expected_columns) - set(df.columns)
unexpected_columns = set(df.columns) - set(expected_columns)

print("Colunas ausentes:", missing_columns)
print("Colunas inesperadas:", unexpected_columns)
print("Valores ausentes:", int(df.isna().sum().sum()))
print("Total de linhas:", len(df))
print("Valores encontrados em Class:", sorted(df["Class"].unique()))

assert not missing_columns, "Existem colunas esperadas ausentes."
assert df["Class"].isin([0, 1]).all(), "Existem rótulos inesperados."
assert df["Class"].notna().all(), "Existem rótulos ausentes."
```

```python
display(df["Time"].describe())

print(
    "Quantidade de valores distintos em Time:",
    df["Time"].nunique()
)

print(
    "Dataset ordenado por Time:",
    df["Time"].is_monotonic_increasing
)
```

```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

# Primeiro, reservamos o conjunto de teste
df_treino_validacao, df_teste = train_test_split(
    df,
    test_size=0.15,
    random_state=RANDOM_STATE,
    stratify=df["Class"],
)

# Depois, separamos treino e validação
df_treino, df_validacao = train_test_split(
    df_treino_validacao,
    test_size=0.15 / 0.85,
    random_state=RANDOM_STATE,
    stratify=df_treino_validacao["Class"],
)

print(f"Treino:     {len(df_treino):,} linhas")
print(f"Validação:  {len(df_validacao):,} linhas")
print(f"Teste:      {len(df_teste):,} linhas")
```

```python
for nome, conjunto in [
    ("Treino", df_treino),
    ("Validação", df_validacao),
    ("Teste", df_teste),
]:
    quantidade_fraudes = int((conjunto["Class"] == 1).sum())
    proporcao_fraudes = conjunto["Class"].mean() * 100

    print(
        f"{nome}: {quantidade_fraudes} fraudes "
        f"({proporcao_fraudes:.4f}%)"
    )
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Separar atributos e variável-alvo
X_treino = df_treino.drop(columns="Class")
y_treino = df_treino["Class"]

X_validacao = df_validacao.drop(columns="Class")
y_validacao = df_validacao["Class"]

X_teste = df_teste.drop(columns="Class")
y_teste = df_teste["Class"]

print("Treino:", X_treino.shape)
print("Validação:", X_validacao.shape)
print("Teste:", X_teste.shape)
```

```python
modelo_logistico = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        (
            "classifier",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=42,
            ),
        ),
    ]
)

modelo_logistico.fit(X_treino, y_treino)

print("Regressão Logística treinada com sucesso.")
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    average_precision_score,
    roc_auc_score,
)

probabilidades = modelo_logistico.predict_proba(X_validacao)[:, 1]
previsoes = (probabilidades >= 0.5).astype(int)

print("Relatório de classificação:")
print(
    classification_report(
        y_validacao,
        previsoes,
        labels=[0, 1],
        target_names=["Normal", "Fraude"],
        digits=4,
        zero_division=0,
    )
)

print("Matriz de confusão:")
print(confusion_matrix(y_validacao, previsoes))

print(
    "Average Precision:",
    round(average_precision_score(y_validacao, probabilidades), 4),
)

print(
    "ROC-AUC:",
    round(roc_auc_score(y_validacao, probabilidades), 4),
)
```

### 6. Ajuste do limiar de decisão

```python

import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    average_precision_score,
)

limiares = [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 0.95]

resultados = []

for limiar in limiares:
    previsoes_limiar = (
        probabilidades >= limiar
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_validacao,
        previsoes_limiar,
        labels=[0, 1],
    ).ravel()

    resultados.append({
        "Limiar": limiar,
        "Precisao": precision_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "Recall": recall_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "F1": f1_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "Falsos positivos": fp,
        "Fraudes detectadas": tp,
        "Fraudes perdidas": fn,
        "Alertas totais": tp + fp,
    })

df_limiares = pd.DataFrame(resultados)

display(
    df_limiares.round(4)
)
```
