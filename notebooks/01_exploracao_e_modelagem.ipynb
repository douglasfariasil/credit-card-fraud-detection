```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

# Primeiro, reservamos o conjunto de teste
df_treino_validacao, df_teste = train_test_split(
    df,
    test_size=0.15,
    random_state=RANDOM_STATE,
    stratify=df["Class"],
)

# Depois, separamos treino e validação
df_treino, df_validacao = train_test_split(
    df_treino_validacao,
    test_size=0.15 / 0.85,
    random_state=RANDOM_STATE,
    stratify=df_treino_validacao["Class"],
)

print(f"Treino:     {len(df_treino):,} linhas")
print(f"Validação:  {len(df_validacao):,} linhas")
print(f"Teste:      {len(df_teste):,} linhas")
```

```python
for nome, conjunto in [
    ("Treino", df_treino),
    ("Validação", df_validacao),
    ("Teste", df_teste),
]:
    quantidade_fraudes = int((conjunto["Class"] == 1).sum())
    proporcao_fraudes = conjunto["Class"].mean() * 100

    print(
        f"{nome}: {quantidade_fraudes} fraudes "
        f"({proporcao_fraudes:.4f}%)"
    )
```

### A — Preparar os dados.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Separar atributos e variável-alvo
X_treino = df_treino.drop(columns="Class")
y_treino = df_treino["Class"]

X_validacao = df_validacao.drop(columns="Class")
y_validacao = df_validacao["Class"]

X_teste = df_teste.drop(columns="Class")
y_teste = df_teste["Class"]

print("Treino:", X_treino.shape)
print("Validação:", X_validacao.shape)
print("Teste:", X_teste.shape)
```

### B — Criar e treinar o modelo.

```python
modelo_logistico = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        (
            "classifier",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=42,
            ),
        ),
    ]
)

modelo_logistico.fit(X_treino, y_treino)

print("Regressão Logística treinada com sucesso.")
```

### C — Avaliar na validação.

```python
from sklearn.metrics import (
    average_precision_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
)

probabilidades = modelo_logistico.predict_proba(X_validacao)[:, 1]
previsoes = (probabilidades >= 0.5).astype(int)

print("Relatório de classificação:")
print(
    classification_report(
        y_validacao,
        previsoes,
        labels=[0, 1],
        target_names=["Normal", "Fraude"],
        digits=4,
        zero_division=0,
    )
)

print("Matriz de confusão:")
print(confusion_matrix(y_validacao, previsoes))

print(
    "Average Precision:",
    round(average_precision_score(y_validacao, probabilidades), 4),
)

print(
    "ROC-AUC:",
    round(roc_auc_score(y_validacao, probabilidades), 4),
)
```

### Ajuste do limiar de decisão

```python

import pandas as pd
from sklearn.metrics import (
    average_precision_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)

limiares = [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 0.95]

resultados = []

for limiar in limiares:
    previsoes_limiar = (
        probabilidades >= limiar
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_validacao,
        previsoes_limiar,
        labels=[0, 1],
    ).ravel()

    resultados.append({
        "Limiar": limiar,
        "Precisao": precision_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "Recall": recall_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "F1": f1_score(
            y_validacao, previsoes_limiar, zero_division=0
        ),
        "Falsos positivos": fp,
        "Fraudes detectadas": tp,
        "Fraudes perdidas": fn,
        "Alertas totais": tp + fp,
    })

df_limiares = pd.DataFrame(resultados)

display(
    df_limiares.round(4)
)
```

### Divisão dos dados sem vazamento por duplicatas

```python
from sklearn.model_selection import StratifiedGroupKFold

# Identifica linhas exatamente iguais, incluindo a classe.
# Linhas iguais recebem o mesmo identificador de grupo.
grupos = df.groupby(
    list(df.columns),
    dropna=False,
    sort=False,
).ngroup().to_numpy()

X = df.drop(columns="Class")
y = df["Class"]

cv = StratifiedGroupKFold(
    n_splits=20,
    shuffle=True,
    random_state=42,
)

# Cada linha recebe o número do fold ao qual foi atribuída.
folds = np.full(len(df), -1, dtype=int)

for fold, (_, indices) in enumerate(cv.split(X, y, groups=grupos)):
    folds[indices] = fold

# Aproximadamente 70% treino, 15% validação e 15% teste.
idx_treino = np.isin(folds, range(14))
idx_validacao = np.isin(folds, range(14, 17))
idx_teste = np.isin(folds, range(17, 20))

df_treino = df.loc[idx_treino].copy()
df_validacao = df.loc[idx_validacao].copy()
df_teste = df.loc[idx_teste].copy()

print("Treino:", df_treino.shape)
print("Validação:", df_validacao.shape)
print("Teste:", df_teste.shape)

for nome, conjunto in [
    ("Treino", df_treino),
    ("Validação", df_validacao),
    ("Teste", df_teste),
]:
    print(
        f"{nome}: "
        f"{conjunto['Class'].value_counts().to_dict()}"
    )
```
